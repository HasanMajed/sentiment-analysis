```python
import re
import emoji
import nltk
import pandas as pd
from nltk.tokenize import word_tokenize
from arabert.preprocess import ArabertPreprocessor

nltk.download('punkt')
```

```python
arabert_prep = ArabertPreprocessor("aubmindlab/bert-base-arabertv02")
```

```python
# Emoji pattern
emoji_pattern = re.compile("[\U00010000-\U0010ffff]", flags=re.UNICODE)

# Tokenizer that preserves emojis
def tokenize_with_emojis(text):
    if pd.isnull(text):
        return []
    
    # Arabic normalization without deleting emojis
    text_clean = text.strip()  # skip arabert.preprocess to keep emojis

    # Split emojis & words using regex
    # This will separate emojis, words, and punctuation
    tokens = re.findall(r'\w+|[^\w\s]', text_clean, re.UNICODE)
    
    # Add emoji tokens separately if not caught
    additional_emojis = emoji_pattern.findall(text_clean)
    tokens += additional_emojis
    tokens = list(dict.fromkeys(tokens))  # remove duplicates while preserving order
    
    return tokens

```

```python
# Make sure your dataframe is already loaded and cleaned
df = pd.read_csv('../../data/complete2_dataset.csv')

df['review'] = df['review'].apply(tokenize_with_emojis)

```
