```python
import pandas as pd

df = pd.read_csv("com_ubercab_reviews.csv")  

# Remove the 'auth_rating' column if it exists
if 'author_rating' in df.columns:
    df = df.drop(columns=['author_rating'])

# Save the cleaned version 
df.to_csv("com_ubercab_reviews.csv", index=False)

df.head()
```

```python
# Remove all newline and carriage return characters
df['review'] = df['review'].str.replace(r'[\r\n]+', ' ', regex=True)

# remove extra spaces again
df['review'] = df['review'].str.strip()
df['review'] = df['review'].str.replace(r'\s+', ' ', regex=True)
df.to_csv('com_ubercab_reviews.csv', index=False)
df.head()
```

```python
import re
import string

# Remove punctuation and symbols from 'review_text'
df['review'] = df['review'].apply(lambda x: re.sub(f"[{re.escape(string.punctuation)}]", "", str(x)))
df['review'] = df['review'].str.replace(r'\s+', ' ', regex=True).str.strip()
df['review'] = df['review'].str.replace(r"[\"']", "", regex=True)
df.to_csv('com_ubercab_reviews.csv', index=False)

df.head()
```

```python
# Convert text in 'review' column to lowercase
df['review'] = df['review'].str.lower()
df.to_csv('com_ubercab_reviews.csv', index=False)

df.head()
```

```python
import nltk
from nltk.corpus import stopwords

# Download stopwords if not already
nltk.download('stopwords')

# Get English stopwords
stop_words = set(stopwords.words('english'))

# Remove stopwords
df['review'] = df['review'].apply(
    lambda x: ' '.join([word for word in str(x).split() if word not in stop_words])
)

df.to_csv('com_ubercab_reviews.csv', index=False)

df.head()
```

```python
import re

def clean_noise(text):
    if pd.isnull(text):
        return ""
    
    # Remove repeated punctuation like !!, ??, ..., ----
    text = re.sub(r'([!?.\-])\1+', r'\1', text)

    # Replace multiple dots or punctuation with a space
    text = re.sub(r'[.]{2,}', ' ', text)

    # Remove multiple spaces
    text = re.sub(r'\s+', ' ', text).strip()

    return text

# Apply to the review_text column
df['review'] = df['review'].apply(clean_noise)
df.to_csv('com_ubercab_reviews.csv', index=False)

display(df)
```

```python
from nltk.stem import WordNetLemmatizer

nltk.download('wordnet')
nltk.download('omw-1.4')

lemmatizer = WordNetLemmatizer()

def apply_lemmatization(text):
    if pd.isnull(text):
        return ""
    return ' '.join([lemmatizer.lemmatize(word) for word in text.split()])
df.to_csv('com_ubercab_reviews.csv', index=False)

display(df)
```

```python
import re

def reduce_repeated_chars(text):
    if pd.isnull(text):
        return ""
    # Replace 3 or more repeated characters with just 2
    return re.sub(r'(.)\1{2,}', r'\1\1', text)
df['review'] = df['review'].apply(reduce_repeated_chars)

display(df)
```

```python
from textblob import TextBlob
import pandas as pd  

def correct_spelling(text):
    if pd.isnull(text):
        return ""
    corrected = TextBlob(text).correct()
    return str(corrected)
df.to_csv('com_ubercab_reviews.csv', index=False)

display(df)
```
