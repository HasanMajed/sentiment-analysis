### Keras Tokenization

```python
import pandas as pd
from tensorflow.keras.preprocessing.text import Tokenizer
from keras.preprocessing.sequence import pad_sequences
from sklearn.preprocessing import LabelEncoder
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, Dropout
```

```python
# Read csv file
df = pd.read_csv('../../data/complete2_dataset.csv')
display(df)
```

```python
# Ensure all entries are strings 
text = df['review'].fillna('').astype(str).tolist()
```

### Encode Sentiment Labels

#### - Convert sentiment labels (Positive, Negative, Neutral) into numerical format.

```python
# Initialize LabelEncoder
label_encoder = LabelEncoder()

# Convert sentiment labels to numerical format using LabelEncoder
df['label_numerical'] = label_encoder.fit_transform(df['label'])
```

### Preprocess Text Data

#### - Tokenize the text data using Keras

```python
# 3. Tokenizer with OOV token
tokenizer = Tokenizer(
    num_words=15_000, 
    filters='', 
    oov_token='<OOV>'
)
# Ensure all reviews are strings (fillna handles NaN)
reviews = df['review'].fillna('').astype(str).tolist()
tokenizer.fit_on_texts(reviews)

# Convert text to sequences
sequences = tokenizer.texts_to_sequences(reviews)
```

#### - Pad Sequences 

```python
# Pad sequences to ensure uniform inout size
# Calculated based on max & avg length of the dataset from EDA file to get the most sutiable max_length 
# to avoid unnecessary padding
max_length = 60
padded = pad_sequences(
    sequences, 
    maxlen=max_length, 
    padding='post', 
    truncating='post'
)
```

```python
# Adding tokenized and padded results to the DataFrame
df['tokenized'] = sequences
df['padded']    = list(padded)
```

```python
# save the code to .pkl file in ../../models
df.to_pickle('../../models/tokenized_data_dl.pkl')
```

```python
# Saveing the results to a new CSV file
df.to_csv('../../data/processed_reviews_dl.csv', encoding='utf-8-sig', index=False)
```
